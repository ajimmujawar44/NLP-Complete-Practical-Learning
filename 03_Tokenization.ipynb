# 1. What is Tokenization?

### Tokenization = splitting text into smaller meaningful units called tokens.

Ex."I love Machine Learning"

Word Tokens:

["I", "love", "Machine", "Learning"]

Each items is a token. 


WHERE : It's the very first structural step in almost every NLP 
task —
        search engines, translators, chatbots all tokenize first.
        
WHY   : Models can't process a raw string as a whole. They need discrete
        units (tokens) to count, embed, and analyze individually.


## 2. Types of Tokenization
You should learn these four types:

                 Tokenization
                      │
       ┌──────────────┼──────────────┐
       ↓              ↓              ↓
   Sentence         Word         Character
   Tokenization    Tokenization   Tokenization
                                     
                      ↓
                Subword Tokenization

| Type      | Example                               | Common use             |
| --------- | ------------------------------------- | ---------------------- |
| Sentence  | `"I like NLP."` → `["I like NLP."]`   | Summarization          |
| Word      | `"I like NLP"` → `["I","like","NLP"]` | Classical NLP          |
| Character | `"NLP"` → `["N","L","P"]`             | Character-level models |
| Subword   | `"playing"` → `["play","##ing"]`      | BERT/Transformers      |


```python
import nltk
from nltk.tokenize import (
    word_tokenize,
    sent_tokenize,
    WhitespaceTokenizer,
    RegexpTokenizer
)


# Download required NLTK data
# nltk.download("punkt")
# nltk.download("punkt_tab")


TEXT = "Dr. Smith went to Washington. He said NLP is amazing! Isn't it great?"


# --------------------------------------------------
# 1. WORD-LEVEL TOKENIZATION
# --------------------------------------------------

def word_level_tokenization(text: str):
    """
    Split text into words and punctuation.
    NLTK handles contractions and punctuation separately.
    """
    return word_tokenize(text)


# --------------------------------------------------
# 2. SENTENCE-LEVEL TOKENIZATION
# --------------------------------------------------

def sentence_level_tokenization(text: str):
    """
    Split text into sentences.
    NLTK can handle cases such as 'Dr.' without
    incorrectly treating it as the end of a sentence.
    """
    return sent_tokenize(text)


# --------------------------------------------------
# 3. WHITESPACE TOKENIZATION
# --------------------------------------------------

def whitespace_tokenization(text: str):
    """
    Split text only on whitespace.

    It is simple and fast, but punctuation remains
    attached to words.
    """
    return WhitespaceTokenizer().tokenize(text)


# --------------------------------------------------
# 4. REGEX WORD TOKENIZATION
# --------------------------------------------------

def regex_word_tokenization(text: str):
    """
    Custom tokenization using regular expressions.

    \\w+ keeps word-like tokens and removes punctuation.
    """
    tokenizer = RegexpTokenizer(r"\w+")
    return tokenizer.tokenize(text)


# --------------------------------------------------
# 5. SUBWORD TOKENIZATION
# --------------------------------------------------

def subword_tokenization_demo():
    """
    Modern NLP models use subword tokenization.

    Examples include:
    - WordPiece
    - BPE
    - SentencePiece

    A word can be divided into smaller pieces.
    """

    print("Subword Tokenization Example (Conceptual):")

    print("  'unbelievable' -> ['un', '##believ', '##able']")

    print(
        "  This helps models handle rare or unknown words "
        "by breaking them into smaller known pieces."
    )


# --------------------------------------------------
# MAIN PROGRAM
# --------------------------------------------------

if __name__ == "__main__":

    print("=" * 60)
    print("ORIGINAL TEXT")
    print("=" * 60)

    print(TEXT)


    print("\n" + "=" * 60)
    print("1. WORD TOKENIZATION")
    print("=" * 60)

    print(word_level_tokenization(TEXT))


    print("\n" + "=" * 60)
    print("2. SENTENCE TOKENIZATION")
    print("=" * 60)

    print(sentence_level_tokenization(TEXT))


    print("\n" + "=" * 60)
    print("3. WHITESPACE TOKENIZATION")
    print("=" * 60)

    print(whitespace_tokenization(TEXT))


    print("\n" + "=" * 60)
    print("4. REGEX WORD TOKENIZATION")
    print("=" * 60)

    print(regex_word_tokenization(TEXT))


    print("\n" + "=" * 60)
    print("5. SUBWORD TOKENIZATION")
    print("=" * 60)

    subword_tokenization_demo()
```

```python
import nltk

# Download tokenizer data
#nltk.download("punkt")
# nltk.download("punkt_tab")

from nltk.tokenize import (
    word_tokenize,
    sent_tokenize
)


text = """
Natural Language Processing is interesting.
I am learning tokenization.
NLP is used in chatbots, search engines and AI.
"""


print("=" * 60)
print("ORIGINAL TEXT")
print("=" * 60)

print(text)


# ------------------------------------------------
# 1. SENTENCE TOKENIZATION
# ------------------------------------------------

sentences = sent_tokenize(text)

print("\n" + "=" * 60)
print("SENTENCE TOKENIZATION")
print("=" * 60)

print(sentences)


# ------------------------------------------------
# 2. WORD TOKENIZATION
# ------------------------------------------------

words = word_tokenize(text)

print("\n" + "=" * 60)
print("WORD TOKENIZATION")
print("=" * 60)

print(words)


# ------------------------------------------------
# 3. NUMBER OF TOKENS
# ------------------------------------------------

print("\n" + "=" * 60)
print("TOKEN COUNT")
print("=" * 60)

print("Number of sentences:", len(sentences))
print("Number of word tokens:", len(words))
```

```python
from transformers import AutoTokenizer


# Load BERT tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)


text = "I am learning Natural Language Processing."


# Tokenize
tokens = tokenizer.tokenize(text)

print("=" * 60)
print("TOKENS")
print("=" * 60)

print(tokens)


# Convert tokens into IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("\n" + "=" * 60)
print("TOKEN IDs")
print("=" * 60)

print(token_ids)


# Complete tokenizer output
result = tokenizer(text)

print("\n" + "=" * 60)
print("TOKENIZER OUTPUT")
print("=" * 60)

print(result)
```
