# Part 1: Introduction to Text Preprocessing

* Text Preprocessing

Text preprocessing is the process of converting raw text into a clean and structured format before giving it to a Machine Learning or Deep Learning model.

In simple words, it removes unnecessary information and keeps only the useful text.

Raw Text
↓

## 02 - TEXT PREPROCESSING
========================

WHAT  : Cleaning raw, messy text before any NLP model can use it —
        lowercasing, removing punctuation/numbers/URLs/HTML tags,
        removing extra whitespace, removing stopwords ("the", "is", "a"...).

WHERE : Every NLP system does this first — search engines, spam filters,
        chatbots, sentiment analyzers all start here.

WHY   : Raw text is noisy and inconsistent ("Good!!!" vs "good" vs "GOOD")
        which would be treated as 3 different tokens by a model. Cleaning
        normalizes text so the model focuses on meaning, not formatting noise.


### Why do we perform Text Preprocessing?

Reason                    | Explanation                                   |
| ------------------------- | --------------------------------------------- |
| Remove unwanted text      | Remove symbols, punctuation, HTML etc.        |
| Improve accuracy          | Better input gives better predictions.        |
| Reduce noise              | Removes unnecessary words.                    |
| Speed up training         | Smaller vocabulary means faster training.     |
| Better feature extraction | Helps Bag of Words and TF-IDF perform better. |

```python
import re
import string


def clean_text(text3):

    # lowercase
    text3 = text3.lower()

    # remove URLs
    text3 = re.sub(
        r"https?://\S+|www\.\S+",
        " ",
        text3
    )

    # remove emails
    text3 = re.sub(
        r"\S+@\S+",
        " ",
        text3
    )

    # remove HTML
    text3 = re.sub(
        r"<.*?>",
        " ",
        text3
    )

    # remove punctuation
    text3 = text3.translate(
        str.maketrans("", "", string.punctuation)
    )

    # remove numbers
    text3 = re.sub(
        r"\d+",
        " ",
        text3
    )

    # remove extra spaces
    text3 = re.sub(
        r"\s+",
        " ",
        text3
    ).strip()

    return text3


text3 = """ HELLO !!! Visit https://google.com.

Email : azimmujaar123@gmail.com.

I bought 2 iphones !!! 😀😍😀🔥 👍"""

cleaned = clean_text(text3)

print("Original:")
print(text3)
print("=" * 60)
print("\nCleaned:")
print(cleaned)
```

```python
# practice Set 2 :
from nltk.corpus import stopwords
STOPWORDS = set(stopwords.words("english"))


def to_lowercase(text: str) -> str:
    return text.lower()


def remove_urls(text: str) -> str:
    return re.sub(r"https?://\S+|www\.\S+", "", text)


def remove_html_tags(text: str) -> str:
    return re.sub(r"<.*?>", "", text)


def remove_punctuation(text: str) -> str:
    return text.translate(str.maketrans("", "", string.punctuation))


def remove_numbers(text: str) -> str:
    return re.sub(r"\d+", "", text)


def remove_extra_whitespace(text: str) -> str:
    return re.sub(r"\s+", " ", text).strip()


def remove_stopwords(text: str) -> str:
    words = text.split()
    filtered = [w for w in words if w not in STOPWORDS]
    return " ".join(filtered)


def full_preprocess_pipeline(text: str) -> str:
    """Runs all cleaning steps in the correct order."""
    text = to_lowercase(text)
    text = remove_urls(text)
    text = remove_html_tags(text)
    text = remove_punctuation(text)
    text = remove_numbers(text)
    text = remove_stopwords(text)
    text = remove_extra_whitespace(text)
    return text


if __name__ == "__main__":
    sample = """
    Check THIS out!!! Visit https://example.com for <b>amazing</b> deals!!
    I bought 3 items in 2024 and I'm SO happy :) #bestdeal
    """
    print("ORIGINAL TEXT:\n", sample)
    print("\nSTEP-BY-STEP CLEANING:")
    print("1. Lowercase        :", to_lowercase(sample).strip())
    print("2. Remove URLs      :", remove_urls(to_lowercase(sample)).strip())
    print("3. Remove Punctuation:", remove_punctuation("Hello!!! How are you??"))
    print("4. Remove Numbers   :", remove_numbers("I have 3 cats and 25 dogs"))
    print("5. Remove Stopwords :", remove_stopwords("this is a sample sentence about nlp"))

    print("\nFINAL CLEANED TEXT:\n", full_preprocess_pipeline(sample))
```
