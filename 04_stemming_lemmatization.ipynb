# 04 - STEMMING & LEMMATIZATION
===============================

WHAT  : Both reduce words to a base/root form so "running", "runs", "ran"
        are treated as the same concept.
        - STEMMING: chops word endings using rules (fast, crude, may not be
          a real word). "studies" -> "studi"
        - LEMMATIZATION: uses vocabulary + grammar (dictionary) to return
          the real base word. "studies" -> "study"

WHERE : Search engines (searching "run" also matches "running"), text
        classification, information retrieval.

WHY   : Reduces vocabulary size and treats words with the same meaning as
        one feature, improving model generalization.


## 1. What is Stemming?

Stemming means reducing a word to a shorter stem by removing prefixes or suffixes using rules.

playing

played

plays

       ↓

play

Useful for:

Search engines

Information retrieval

Text classification

Document matching

Some traditional NLP systems

## Porter Stemmer 

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

print(stemmer.stem("playing"))
print(stemmer.stem("played"))
print(stemmer.stem("studies"))
```

```python
stemmer = PorterStemmer()
print(stemmer.stem("Easily"))
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("cars"))
```

```python
lemmatizer.lemmatize("better", pos="a")
```

## 3. Complete Example 

```python
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

words = [
    "playing",
    "played",
    "studies",
    "cars",
    "mice"
]

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

for word in words:

    stem = stemmer.stem(word)

    lemma = lemmatizer.lemmatize(word)

    print(
        f"{word:10} "
        f"Stem: {stem:10} "
        f"Lemma: {lemma}"
    )
```

## 4. Complete Code Practice 

```python

import nltk

from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
from nltk.stem import LancasterStemmer
from nltk.stem import WordNetLemmatizer


# Download required data
nltk.download("wordnet")
nltk.download("omw-1.4")


# Words for testing
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "running",
    "easily",
    "fairly",
    "connected",
    "connection"
]


print("=" * 60)
print("STEMMING AND LEMMATIZATION")
print("=" * 60)


# --------------------------------------------------
# 1. Porter Stemmer
# --------------------------------------------------

porter = PorterStemmer()

print("\nPorter Stemmer")
print("-" * 30)

for word in words:
    print(word, "->", porter.stem(word))


# --------------------------------------------------
# 2. Snowball Stemmer
# --------------------------------------------------

snowball = SnowballStemmer("english")

print("\nSnowball Stemmer")
print("-" * 30)

for word in words:
    print(word, "->", snowball.stem(word))


# --------------------------------------------------
# 3. Lancaster Stemmer
# --------------------------------------------------

lancaster = LancasterStemmer()

print("\nLancaster Stemmer")
print("-" * 30)

for word in words:
    print(word, "->", lancaster.stem(word))


# --------------------------------------------------
# 4. WordNet Lemmatizer
# --------------------------------------------------

lemmatizer = WordNetLemmatizer()

print("\nWordNet Lemmatizer")
print("-" * 30)

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))


# --------------------------------------------------
# 5. Lemmatization with POS
# --------------------------------------------------

print("\nLemmatization with POS")
print("-" * 30)

pos_words = [
    ("playing", "v"),
    ("played", "v"),
    ("running", "v"),
    ("studies", "v"),
    ("better", "a"),
    ("cars", "n"),
    ("mice", "n")
]

for word, pos in pos_words:

    result = lemmatizer.lemmatize(word, pos=pos)

    print(word, "->", result)

```
