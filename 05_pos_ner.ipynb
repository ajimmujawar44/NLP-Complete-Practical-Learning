05 - POS TAGGING & NAMED ENTITY RECOGNITION (NER)
===================================================

WHAT  : - POS (Part-of-Speech) Tagging: labels every word with its grammar
          role — Noun, Verb, Adjective, etc. "NLP is fun" -> NLP/NOUN is/VERB fun/ADJ
        - NER (Named Entity Recognition): finds real-world entities in text
          — People, Organizations, Locations, Dates, Money.

WHERE : - POS -> grammar checkers, question answering, syntactic parsing.
        - NER -> resume parsing (extract names/companies), news categorization,
          chatbots extracting slots ("book a flight to Paris on Friday").
          
WHY   : Raw tokens alone don't tell you a word's ROLE or IDENTITY. POS/NER
        add structure & meaning needed for downstream understanding.


This task combines Part-of-Speech (POS) Tagging and Named Entity Recognition (NER) because both are related to understanding the role and meaning of words in a sentence.

## POS Tagging

POS tells us what type of word each word is.

Example:

Ajim     → PROPN


is       → AUX

learning → VERB

NLP      → PROPN



## NER — Named Entity Recognition

NER finds important named entities in text.

Example:

Ajim lives in India and works with Microsoft.

* NER can identify:

Ajim       → PERSON

India      → GPE

Microsoft  → ORG

```python
import spacy


# Load English language model
nlp = spacy.load("en_core_web_sm")


text = """
Ajim is learning Natural Language Processing in India.
He is using Python and spaCy to build NLP applications.
"""


# Process the text
doc = nlp(text)


# ==================================================
# 1. PART-OF-SPEECH TAGGING
# ==================================================

print("=" * 60)
print("PART-OF-SPEECH TAGGING")
print("=" * 60)

for token in doc:

    print(token.text, "->", token.pos_)


# ==================================================
# 2. POS TAG + DETAILED INFORMATION
# ==================================================

print("\n" + "=" * 60)
print("POS TAG DETAILS")
print("=" * 60)

for token in doc:

    print(
        token.text,
        "->",
        token.pos_,
        "->",
        token.tag_
    )


# ==================================================
# 3. NAMED ENTITY RECOGNITION
# ==================================================

print("\n" + "=" * 60)
print("NAMED ENTITY RECOGNITION")
print("=" * 60)

for entity in doc.ents:

    print(
        entity.text,
        "->",
        entity.label_
    )


# ==================================================
# 4. ENTITY EXPLANATION
# ==================================================

print("\n" + "=" * 60)
print("ENTITY EXPLANATION")
print("=" * 60)

for entity in doc.ents:

    print(
        entity.text,
        "->",
        entity.label_,
        "->",
        spacy.explain(entity.label_)
    )
```

```python

import nltk
#nltk.download("averaged_perceptron_tagger", quiet=True)
#nltk.download("averaged_perception_tagger_eng", quiet=True)
#nltk.download("maxent_ne_chunkar", quiet=True)
#nltk.dpwnload("Maxent_ne_chunkar_tab", quiet='True')
#nltk.doenload("words",quiet=True)
#nltk.doenload("punk_lab",quiet=True)
#nltk.download("punkt_tab", quiet=True)

from nltk import pos_tag, word_tokenize, ne_chunk

TEXT="Elon Musk founded SpaceX in California in 2002 and it is now worth billions."


def pos_tagging_nltk(text: str):
    """
    Tags each token with its part of speech.
    Common tags: NNP=Proper Noun, VBD=Verb Past Tense, JJ=Adjective, IN=Preposition
    """
    tokens = word_tokenize(text)
    return pos_tag(tokens)


def named_entity_recognition_nltk(text: str):
    """Chunks POS-tagged tokens into named entity trees (PERSON, ORGANIZATION, GPE)."""
    tagged = pos_tag(word_tokenize(text))
    tree = ne_chunk(tagged)
    entities = []
    for chunk in tree:
        if hasattr(chunk, "label"):
            entity_text = " ".join(c[0] for c in chunk)
            entities.append((entity_text, chunk.label()))
    return entities


def pos_ner_with_spacy(text: str):
    """
    spaCy gives cleaner, production-quality POS + NER in one pass.
    Requires: python -m spacy download en_core_web_sm
    """
    try:
        import spacy
        nlp = spacy.load("en_core_web_sm")
        doc = nlp(text)
        print("\n[spaCy] POS Tags:")
        for token in doc:
            print(f"  {token.text:<12} -> {token.pos_} ({token.tag_})")
        print("\n[spaCy] Named Entities:")
        for ent in doc.ents:
            print(f"  {ent.text:<15} -> {ent.label_}")
    except OSError:
        print("\n[spaCy] Model not found. Run: python -m spacy download en_core_web_sm")
    except ImportError:
        print("\n[spaCy] Not installed. Run: pip install spacy")


if __name__ == "__main__":
    print("TEXT:", TEXT)

    print("\n[NLTK] POS Tags:")
    for word, tag in pos_tagging_nltk(TEXT):
        print(f"  {word:<12} -> {tag}")

    print("\n[NLTK] Named Entities:")
    for ent_text, label in named_entity_recognition_nltk(TEXT):
        print(f"  {ent_text:<15} -> {label}")

    pos_ner_with_spacy(TEXT)
```

### POS tagging a simple senttence

```python
nlp  = spacy.load("en_core_web_sm")

text = "the boy is playing football."

doc = nlp(text)

print("=" * 50)
print("POS TAGGING")
print("="* 50)

for token in doc :
    print(token.text, "->", token.pos_)
```

### Find only nouns

```python
text = "Python is a popular programming language."

doc = nlp(text)

print("NOUNS:")

for token in doc:

    if token.pos_ == "NOUN":
        print(token.text)
```

### Find only verbs

```python
text = "Python is easy and developers use it to build applications."

doc = nlp(text)

print("VERBS:")

for token in doc:

    if token.pos_ == "VERB":
        print(token.text)
```

### Extract Organization

```python
text = """
Microsoft is a technology company.
Google develops many software products.
Apple makes computers and phones.
"""

doc = nlp(text)

print("ORGANIZATIONS:")

for entity in doc.ents:

    if entity.label_ == "ORG":
        print(entity.text)
```

### Count POS Tags

```python
from collections import Counter

text = """
Python is a popular programming language.
Python is easy to learn and useful for data science.
"""

doc = nlp(text)

pos_count = Counter()

for token in doc:

    if not token.is_punct:
        pos_count[token.pos_] += 1

print(pos_count)

```
