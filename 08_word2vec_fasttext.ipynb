# 08 - WORD2VEC & FASTTEXT 

* WHAT  : Algorithms that LEARN word embeddings from a text corpus by
        predicting context.
        - Word2Vec (Skip-gram/CBOW): learns from whole words only.
        - FastText: learns from CHARACTER n-grams inside words, so it can
          handle misspellings and words never seen during training
          (out-of-vocabulary / OOV words).

* WHERE : Domain-specific search (legal, medical text where generic
        embeddings don't fit), recommendation engines, any place you have
        your OWN large text corpus and want embeddings tailored to it.

* WHY   : Pretrained embeddings (Google News Word2Vec) may not understand
        YOUR domain's jargon. Training your own on your own corpus (e.g.
        product reviews, medical notes) gives more relevant vectors.

* Requires: pip install gensim

## what is Word2Vec?

Word2Vec is a technique for learning numerical representations of words from their surrounding context.

main idea is: Words that appear in similar contexts tend to have similar meanings.

ex.-The cat drinks milk.
    The dog drinks milk.

       cat
      /   \
   The    drinks
            |
           milk
* word 2 vec architures:

Word2Vec

   |
   ├── CBOW
   |
   └── Skip-gram

## CBOW : Continuous Bag of Words

It uses surrounding words to predict the target word.

EX. The cat drinks milk

Context words

     ↓

   Predict

     ↓

 Target word

 ## Skip gram 

Skip-gram works in the opposite direction.

It takes a target word and predicts surrounding words.

Example:

cat

 ↓
Skip-gram

 ↓

The
drinks

## CBOW vs Skip-gram

| Method    | Input         | Predicts      |
| --------- | ------------- | ------------- |
| CBOW      | Context words | Target word   |
| Skip-gram | Target word   | Context words |

## First Word2Vec Example 

```python
from gensim.models import Word2Vec


sentences = [
    ["i", "love", "python"],
    ["i", "love", "nlp"],
    ["python", "is", "easy"],
    ["nlp", "is", "interesting"],
    ["python", "is", "useful"]
]


model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1
)


print("Vocabulary:")
print(model.wv.index_to_key)
```

```python
# Get a word vector :

vector = model.wv["python"]

print(vector)
```

```python
# Check Vector shape :

print(vector.shape)
```

```python
# Finding Simillr words :

similar_words = model.wv.most_similar("python")

print(similar_words)
```

# What is FastText?

FastText was developed by Facebook/Meta's AI research group.

def.= Instead of treating a word as one indivisible unit,         FastText also learns from character/subword pieces.

ex. 

playing

 ↓
play

lay

ayin

ying
...

The exact subword generation depends on FastText's settings.

* This makes it particularly useful for:

Rare words

Morphologically rich languages

Misspellings

New/unseen words

Words with shared prefixes/suffixes

## Word2Vec vs FastText :

| Feature                                | Word2Vec     | FastText  |
| -------------------------------------- | ------------ | --------- |
| Word-level vectors                     | ✅            | ✅         |
| Uses surrounding context               | ✅            | ✅         |
| Uses subwords                          | ❌            | ✅         |
| Handles rare words better              | Limited      | Better    |
| Can construct vectors for unseen words | Generally no | Often yes |
| CBOW                                   | ✅            | ✅         |
| Skip-gram                              | ✅            | ✅         |

## Complete practice example on word2vec and fastText:



```python
# Example 1: 

from gensim.models import Word2Vec
from gensim.models import FastText


print("=" * 60)
print("WORD2VEC AND FASTTEXT")
print("=" * 60)


# --------------------------------------------------
# Training sentences
# --------------------------------------------------

sentences = [
    ["the", "cat", "drinks", "milk"],
    ["the", "dog", "drinks", "milk"],
    ["the", "cat", "eats", "food"],
    ["the", "dog", "eats", "food"],
    ["the", "cat", "likes", "fish"],
    ["the", "dog", "likes", "meat"],

    ["python", "is", "a", "programming", "language"],
    ["python", "is", "easy", "to", "learn"],
    ["python", "is", "powerful", "and", "useful"],

    ["machine", "learning", "uses", "python"],
    ["data", "science", "uses", "python"]
]


# ==================================================
# 1. WORD2VEC
# ==================================================

print("\n" + "=" * 60)
print("WORD2VEC")
print("=" * 60)


word2vec = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1
)


# --------------------------------------------------
# Vocabulary
# --------------------------------------------------

print("\nWord2Vec Vocabulary:")

print(
    word2vec.wv.index_to_key
)


# --------------------------------------------------
# Word Vector
# --------------------------------------------------

print("\nVector for Python:")

print(
    word2vec.wv["python"]
)


# --------------------------------------------------
# Vector Shape
# --------------------------------------------------

print("\nVector Shape:")

print(
    word2vec.wv["python"].shape
)


# --------------------------------------------------
# Similar Words
# --------------------------------------------------

print("\nWords similar to Python:")

print(
    word2vec.wv.most_similar("python")
)


# --------------------------------------------------
# Word Similarity
# --------------------------------------------------

print("\nSimilarity between cat and dog:")

print(
    word2vec.wv.similarity(
        "cat",
        "dog"
    )
)


# ==================================================
# 2. FASTTEXT
# ==================================================

print("\n" + "=" * 60)
print("FASTTEXT")
print("=" * 60)


fasttext = FastText(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1
)


# --------------------------------------------------
# Vocabulary
# --------------------------------------------------

print("\nFastText Vocabulary:")

print(
    fasttext.wv.index_to_key
)


# --------------------------------------------------
# Word Vector
# --------------------------------------------------

print("\nVector for Python:")

print(
    fasttext.wv["python"]
)


# --------------------------------------------------
# Vector Shape
# --------------------------------------------------

print("\nVector Shape:")

print(
    fasttext.wv["python"].shape
)


# --------------------------------------------------
# Similar Words
# --------------------------------------------------

print("\nWords similar to Python:")

print(
    fasttext.wv.most_similar("python")
)


# --------------------------------------------------
# Word Similarity
# --------------------------------------------------

print("\nSimilarity between cat and dog:")

print(
    fasttext.wv.similarity(
        "cat",
        "dog"
    )
)


# --------------------------------------------------
# Unknown / New Word
# --------------------------------------------------

print("\nVector for a new word:")

print(
    fasttext.wv["pythonic"]
)
```

```python
# Example 2:

from gensim.models import Word2Vec, FastText

# A tiny toy corpus — in real projects this would be thousands of sentences
CORPUS = [
    "I love natural language processing".split(),
    "Machine learning is a subset of artificial intelligence".split(),
    "Deep learning uses neural networks".split(),
    "Natural language processing uses machine learning".split(),
    "I love deep learning and artificial intelligence".split(),
    "Neural networks power modern NLP models".split(),
]


def train_word2vec(corpus):
    """
    vector_size : dimensions of each word vector
    window      : how many neighboring words to consider as 'context'
    min_count   : ignore words that appear fewer than this many times
    sg=1        : use Skip-gram (predict context from word) — good for small data
    """
    model = Word2Vec(sentences=corpus, vector_size=50, window=3, min_count=1, sg=1, epochs=100)
    return model


def train_fasttext(corpus):
    """Same idea as Word2Vec but also learns from subword character n-grams."""
    model = FastText(sentences=corpus, vector_size=50, window=3, min_count=1, epochs=100)
    return model


if __name__ == "__main__":
    print("Training Word2Vec on a tiny toy corpus...")
    w2v_model = train_word2vec(CORPUS)
    print("Vocabulary:", list(w2v_model.wv.index_to_key))

    print("\nMost similar words to 'learning' (Word2Vec):")
    for word, score in w2v_model.wv.most_similar("learning", topn=3):
        print(f"  {word:<15} score={score:.3f}")

    print("\nTraining FastText on the same corpus...")
    ft_model = train_fasttext(CORPUS)

    print("\nFastText can handle a word it NEVER saw during training,")
    print("e.g. 'learnings' (misspelling/variant of 'learning'), because")
    print("it builds the vector from character n-grams:")
    try:
        vec = ft_model.wv["learnings"]
        print(f"  Vector generated successfully, shape={vec.shape}")
        print("  Most similar to 'learnings':", ft_model.wv.most_similar("learnings", topn=3))
    except KeyError as e:
        print("  Error:", e)

    print("\nNOTE: With only 6 toy sentences, similarity scores are not very")
    print("meaningful. In real projects, train on thousands+ of sentences,")
    print("or better — use pretrained models (see module 07) for general use.")
```
