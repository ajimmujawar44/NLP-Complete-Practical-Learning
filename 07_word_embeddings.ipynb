# 07 - WORD EMBEDDINGS



* WHAT  : Dense numeric vector representations of words where similar
        meaning words are CLOSE together in vector space.
        e.g. vector("king") - vector("man") + vector("woman") ≈ vector("queen")
        Unlike BoW/TF-IDF (sparse, no meaning captured), embeddings are
        DENSE and capture SEMANTIC relationships.

* WHERE : Recommendation systems, search relevance, chatbots, translation,
        and as the INPUT layer for almost every deep learning NLP model.

* WHY   : TF-IDF has no idea that "good" and "great" are related — they're
        just two different columns. Embeddings place them near each other
        in vector space, letting models generalize/understand synonyms.



# 1. What are World Embeddings ?

A word embedding represents a word as a dense numerical vector that can learn useful relationships and semantic information.

* Bag of Words

TF-IDF

word embedding represent each word as a vector

King   → [0.21, 0.83, 0.14, ...]

Queen  → [0.25, 0.79, 0.18, ...]

Man    → [0.19, 0.72, 0.12, ...]

Woman  → [0.23, 0.75, 0.16, ...]

Words with similar meanings tend to have similar vector representations.

## Why did we need word Embeddings?

* Machine-learning and neural-network models need numerical input, and embeddings provide a more meaningful representation than simple word counts.

Bags of Words

King → [1, 0, 0, 0]

Queen → [0, 1, 0, 0]

It treats them as completely separate words.
## Word Embeddings

King  → vector

Queen → similar vector

The vectors can capture relationships between words.

## One-Hot Encoding vs Word Embedding

Before embeddings, we should understand One-Hot Encoding.

Suppose:
["cat", "dog", "king", "queen"]

one-hot-encder representation :

cat   → [1, 0, 0, 0]

dog   → [0, 1, 0, 0]

king  → [0, 0, 1, 0]

queen → [0, 0, 0, 1]

## Simple One-Hot Example 


```python
words = ["cat", "dog", "king", "queen"]

print("One-Hot Encoding")

for i, word in enumerate(words):

    vector = [0] * len(words)

    vector[i] = 1

    print(word, "->", vector)
```

## Complete simple embedding example

```python
import numpy as np

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding


print("=" * 60)
print("WORD EMBEDDINGS")
print("=" * 60)


# --------------------------------------------------
# Sample sentences
# --------------------------------------------------

sentences = [
    "I love Python",
    "I love NLP",
    "Python is easy",
    "NLP is interesting"
]


# --------------------------------------------------
# 1. Convert words into numbers
# --------------------------------------------------

tokenizer = Tokenizer()

tokenizer.fit_on_texts(sentences)

print("\nWord Index:")
print(tokenizer.word_index)


# --------------------------------------------------
# 2. Convert sentences into sequences
# --------------------------------------------------

sequences = tokenizer.texts_to_sequences(sentences)

print("\nSequences:")
print(sequences)


# --------------------------------------------------
# 3. Find vocabulary size
# --------------------------------------------------

vocab_size = len(tokenizer.word_index) + 1

print("\nVocabulary Size:")
print(vocab_size)


# --------------------------------------------------
# 4. Make all sentences same length
# --------------------------------------------------

padded_sequences = pad_sequences(
    sequences,
    padding="post"
)

print("\nPadded Sequences:")
print(padded_sequences)


# --------------------------------------------------
# 5. Create Embedding Layer
# --------------------------------------------------

embedding_size = 5

model = Sequential()

model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_size
    )
)

model.build(
    input_shape=(None, padded_sequences.shape[1])
)


# --------------------------------------------------
# 6. Display model
# --------------------------------------------------

print("\nModel:")
model.summary()


# --------------------------------------------------
# 7. Generate embeddings
# --------------------------------------------------

embeddings = model.predict(
    padded_sequences,
    verbose=0
)

print("\nEmbedding Shape:")
print(embeddings.shape)


# --------------------------------------------------
# 8. Display embedding
# --------------------------------------------------

print("\nEmbedding Values:")
print(embeddings)
```
