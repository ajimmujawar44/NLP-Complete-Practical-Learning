## 06 - TEXT FEATURE EXTRACTION: BOW, TF-IDF, N-GRAMS
=====================================================

* WHAT  : Techniques to convert text into NUMBERS (vectors) so ML models
        (which only understand math) can use them.
        - Bag of Words (BoW): counts how many times each word appears.
        - TF-IDF: weighs words by importance (frequent in THIS doc,
          rare across ALL docs = important).
        - N-grams: captures word sequences (bigrams="not good" as ONE
          unit, instead of losing meaning by treating "not" and "good"
          separately).

* WHERE : Spam detection, sentiment analysis, document search/ranking
        (this is literally how early Google search worked).

* WHY   : "The cat sat" is meaningless to a model until it's a vector like
        [0,1,0,1,1,...]. TF-IDF specifically fixes BoW's flaw of over-
        weighting common-but-unimportant words like "the".

## Simple N-Grams example

```python
text = "I LOVE Learning NLP"

words  = text.split()

print("words:")
print(words)

print("\nUnigrams:")

for word in words:
    print(word)

print("\nBigrams:")

for i in range(len(words)-1):
    print(words[i], words[i+1])

print("\nTrigrams:")

for i in range(len(words)-2):
    print(words[i], words[i+1], words[i+2])
```

### CountVectorizr

insted of creating BoW manually, scikit-learn gives us:

```python
from sklearn.feature_extraction.text import CountVectorizer


sentences = [
    "I love Python",
    "I love NLP",
    "Python is easy"
]


vectorizer = CountVectorizer()

X = vectorizer.fit_transform(sentences)


print("Vocabulary:")
print(vectorizer.get_feature_names_out())


print("\nBag of Words:")
print(X.toarray())
```

### Bigram with CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer


sentences = [
    "I love Python",
    "I love NLP",
    "Python is easy"
]


vectorizer = CountVectorizer(
    ngram_range=(1, 2)
)

X = vectorizer.fit_transform(sentences)


print("Features:")
print(vectorizer.get_feature_names_out())


print("\nFeature Matrix:")
print(X.toarray())
```

```python
# Only Biagrams:

vectorizer = CountVectorizer(
    ngram_range=(2, 2)
)

X = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
```

```python
# Only Trigrams:

vectorizer = CountVectorizer(
    ngram_range=(3, 3)
)

X = vectorizer.fit_transform(sentences)

print(vectorizer.get_feature_names_out())
```

### TfidfVectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer


sentences = [
    "I love Python",
    "I love NLP",
    "Python is easy"
]


vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(sentences)


print("Features:")
print(vectorizer.get_feature_names_out())


print("\nTF-IDF Matrix:")
print(X.toarray())
```

### TF-IDF with Bigrams

```python
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2)
)

X = vectorizer.fit_transform(sentences)

print("Features:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF:")
print(X.toarray())
```

### Complete Learner-style file

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer


print("=" * 60)
print("N-GRAMS, BAG OF WORDS AND TF-IDF")
print("=" * 60)


# --------------------------------------------------
# Sample sentences
# --------------------------------------------------

sentences = [
    "I love Python",
    "I love NLP",
    "Python is easy"
]


# --------------------------------------------------
# 1. N-GRAMS
# --------------------------------------------------

text = "I love learning NLP"

words = text.split()

print("\nUnigrams:")

for word in words:
    print(word)


print("\nBigrams:")

for i in range(len(words) - 1):

    print(words[i], words[i + 1])


print("\nTrigrams:")

for i in range(len(words) - 2):

    print(words[i], words[i + 1], words[i + 2])


# --------------------------------------------------
# 2. BAG OF WORDS
# --------------------------------------------------

print("\n" + "=" * 60)
print("BAG OF WORDS")
print("=" * 60)

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(sentences)

print("\nVocabulary:")
print(vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(X.toarray())


# --------------------------------------------------
# 3. BAG OF WORDS WITH BIGRAMS
# --------------------------------------------------

print("\n" + "=" * 60)
print("BAG OF WORDS WITH BIGRAMS")
print("=" * 60)

vectorizer = CountVectorizer(
    ngram_range=(1, 2)
)

X = vectorizer.fit_transform(sentences)

print("\nFeatures:")
print(vectorizer.get_feature_names_out())

print("\nMatrix:")
print(X.toarray())


# --------------------------------------------------
# 4. TF-IDF
# --------------------------------------------------

print("\n" + "=" * 60)
print("TF-IDF")
print("=" * 60)

tfidf = TfidfVectorizer()

X = tfidf.fit_transform(sentences)

print("\nFeatures:")
print(tfidf.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(X.toarray())


# --------------------------------------------------
# 5. TF-IDF WITH BIGRAMS
# --------------------------------------------------

print("\n" + "=" * 60)
print("TF-IDF WITH BIGRAMS")
print("=" * 60)

tfidf = TfidfVectorizer(
    ngram_range=(1, 2)
)

X = tfidf.fit_transform(sentences)

print("\nFeatures:")
print(tfidf.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(X.toarray())


# --------------------------------------------------
# 6. NEW SENTENCE
# --------------------------------------------------

print("\n" + "=" * 60)
print("NEW SENTENCE")
print("=" * 60)

new_sentence = ["I love Python"]

X_new = tfidf.transform(new_sentence)

print("\nNew sentence:")
print(new_sentence)

print("\nConverted into numbers:")
print(X_new.toarray())
```
